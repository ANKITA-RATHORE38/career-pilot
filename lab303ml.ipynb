```python
import requests

# Function to get latitude and longitude from city name
def get_coordinates(city):
    try:
        url = f"https://geocoding-api.open-meteo.com/v1/search?name={city}&count=1"

        response = requests.get(url)
        response.raise_for_status()

        data = response.json()

        if "results" in data:
            result = data["results"][0]
            return result["latitude"], result["longitude"], result["name"]
        else:
            return None, None, None

    except requests.exceptions.RequestException as e:
        print("Error:", e)
        return None, None, None


# Function to get weather data
def get_weather(lat, lon):
    try:
        url = (
            f"https://api.open-meteo.com/v1/forecast?"
            f"latitude={lat}&longitude={lon}"
            f"&hourly=temperature_2m,rain"
        )

        response = requests.get(url)
        response.raise_for_status()

        return response.json()

    except requests.exceptions.RequestException as e:
        print("Error:", e)
        return None


# Function to display weather data
def display_weather(data, city):
    if data:
        print(f"\n------ Weather Forecast for {city} ------")

        time = data["hourly"]["time"]
        temp = data["hourly"]["temperature_2m"]
        rain = data["hourly"]["rain"]

        # Display first 5 records
        for i in range(5):
            print("Time        :", time[i])
            print("Temperature :", temp[i], "°C")
            print("Rain        :", rain[i], "mm")
            print("-----------------------------")
    else:
        print("No weather data available.")


# Main function
def main():
    while True:
        try:
            print("\n===== Weather Forecasting System =====")

            city = input("Enter City Name: ").strip()

            lat, lon, city_name = get_coordinates(city)

            if lat is None:
                print("City not found.")
                continue

            weather = get_weather(lat, lon)
            display_weather(weather, city_name)

            choice = input("Check another location? (yes/no): ").lower()

            if choice != "yes":
                print("Program Ended.")
                break

        except Exception as e:
            print("Unexpected Error:", e)


# Function Call
if __name__ == "__main__":
    main()
```
